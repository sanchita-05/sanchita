```python
#Polymorphism
class Shape:
  def area(self):
    pass
  def perimeter(self):
    pass


class Square(Shape):
  def __init__(self,side):
   self.side=side
  def area(self):
    return self.side**2
  def perimeter(self):
    return 4*self.side

class Rectangle(Shape):
  def __init__(self,width,height):
    self.width=width
    self.height=height
  def area(self):
    return self.width*self.height
  def perimeter(self):
    return 2*(self.width+self.height)

class Circle(Shape):
  def __init__(self,radius):
      self.radius=radius
  def area(self):
      return 3.14 * self.radius ** 2
  def perimeter(self):
      return 3.14 * self.radius * 2

#creating objects of shape classes
square=Square(4)
rectangle=Rectangle(10,20)
circle=Circle(5)

#Calculating area and perimeter
print("square area:",square.area())
print("square perimeter:",square.perimeter())
print("rectangle area:",rectangle.area())
print("rectangle perimeter:",rectangle.perimeter())
print("circle area:",circle.area())
print("circle perimeter:",circle.perimeter())

#calling methods
square.area()
square.perimeter()
rectangle.area()
rectangle.perimeter()
circle.area()
circle.perimeter()
```

```python
# Polymorphism of vehicle
class Vehicle:
 def __init__(self,name,color,price):
    self.name=name
    self.color=color
    self.price=price
    def move(self):
     print("move!")

class Car(Vehicle):
  def move(self):
    print("drive!")

class Truck(Vehicle):
  def move(self):
    print("load!")

car1 = Car("toyota","red","300")
truck1 = Truck("Ford","black","500")

for x in (car1,truck1):
  print(x.name)
  print(x.color)
  print(x.price)
  x.move()


```

```python
#Inheritance

class Vehicle:
  def __init__(self,name,color,price):
    self.name=name
    self.color=color
    self.price=price

  def start_engine(self):
    print("starting engine")


class Car(Vehicle):
   def __init__(self,name,color,price,doors):
    super().__init__(name,color,price)
    self.doors=doors

   def start_engine(self):
    print("starting engine")
   def drive_engine(self):
    print("drive engine")

vehicle=Vehicle("toyota","red",300)
car=Car("toyota","red",300,4)
    #print
vehicle.start_engine()
#vehicle.drive_engine gives error bcoz not defined in that class
car.start_engine()
car.drive_engine()
```
